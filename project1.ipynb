```python
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Python Version:", sys.version)
print("NumPy Version:", np.__version__)
print("Pandas Version:", pd.__version__)

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(6, 4))
plt.plot(x, y, label="sin(x)")
plt.title("Kernel Test")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.grid(True)
plt.show()

print("✅ Kernel is working correctly!")
```

```python
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

print("=" * 50)
print("KERNEL TEST")
print("=" * 50)

print("Python:", sys.version)
print("Executable:", sys.executable)
print("Current Directory:", os.getcwd())

# NumPy Test
arr = np.random.randint(1, 100, (5, 5))
print("\nNumPy Array:")
print(arr)

# Pandas Test
df = pd.DataFrame(arr, columns=["A", "B", "C", "D", "E"])
print("\nPandas DataFrame:")
print(df)

# Statistics
print("\nSummary:")
print(df.describe())

# Plot Test
plt.figure(figsize=(7,4))
plt.plot(df["A"], marker="o", label="Column A")
plt.plot(df["B"], marker="s", label="Column B")
plt.title("Kernel Stress Test")
plt.legend()
plt.grid(True)
plt.show()

print("\n✅ All tests passed successfully!")
```
